```python
# This is some random source I found locally, I dont want to fetch from s3 now. Could be implemented here later.
EDISON_DIR = "/Users/makfazlic/Desktop/YUGIOH_FILES/10k_replays_goat_edi/replays-EDISON"
GOAT_DIR = "/Users/makfazlic/Desktop/YUGIOH_FILES/10k_replays_goat_edi/replays-GOAT"
```

```python
import os
import random
import shutil
```

```python
edison_files = os.listdir(EDISON_DIR)
goat_files = os.listdir(GOAT_DIR)

random.shuffle(edison_files)
random.shuffle(goat_files)

subset_size = 100
edison_files = edison_files[:subset_size]
goat_files = goat_files[:subset_size]


EDISON_DEST_DIR = "./100_edison_games"
GOAT_DEST_DIR = "./100_goat_games"

for file in os.listdir(EDISON_DEST_DIR):
    os.remove(os.path.join(EDISON_DEST_DIR, file))
for file in os.listdir(GOAT_DEST_DIR):
    os.remove(os.path.join(GOAT_DEST_DIR, file))

for i, file in enumerate(edison_files):
    shutil.copy(os.path.join(EDISON_DIR, file), os.path.join(EDISON_DEST_DIR, f"{i+1}.json"))

for i, file in enumerate(goat_files):
    shutil.copy(os.path.join(GOAT_DIR, file), os.path.join(GOAT_DEST_DIR, f"{i+1}.json"))

print(f"Number of files in {EDISON_DEST_DIR}: {len(os.listdir(EDISON_DEST_DIR))}")
print(f"Number of files in {GOAT_DEST_DIR}: {len(os.listdir(GOAT_DEST_DIR))}")
```
