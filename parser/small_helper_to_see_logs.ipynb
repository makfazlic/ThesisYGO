```python
import json, os
from tqdm import tqdm
```

```python
def get_log_text(play: dict) -> str:
    log = play.get("log") or {}
    if not isinstance(log, dict):
        return ""
    return (log.get("private_log") or log.get("public_log") or "").strip()
```

```python
logs = []
```

```python
plays_set = set()

files = os.listdir("../replay_manager/100_goat_games")
for file in tqdm(files):
    with open(os.path.join("../replay_manager/100_goat_games", file), "r") as f:
        data = json.load(f)
        plays = data.get("plays", [])
        for play in plays:
            play_type = play.get("play")
            if play_type:
                plays_set.add(play_type)

```

```python
len(plays_set)
for i, play in enumerate(sorted(plays_set)):
    print(i, play)
```

```python
files = os.listdir("../replay_manager/100_goat_games")
for file in tqdm(files):
    with open(os.path.join("../replay_manager/100_goat_games", file), "r") as f:
        data = json.load(f)
        plays = data.get("plays", [])
        for play in plays:
            play_type = play.get("play")
            if play_type == "To ST":
                log = get_log_text(play)
                logs.append("Flip log: " + log)
```

```python
# randomly sample 100 logs to see
import random
sampled_logs = random.sample(logs, min(100, len(logs)))
for log in sampled_logs:
    print("log=", log)
```
