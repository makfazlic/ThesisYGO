```python
def get_log_text(play: dict) -> str:
    log = play.get("log") or {}
    if not isinstance(log, dict):
        return ""
    return (log.get("private_log") or log.get("public_log") or "").strip()
```

```python
import json
import re
from pathlib import Path
from collections import Counter, defaultdict

# Find the repository from either its root or a notebook in parser/.
HERE = Path.cwd().resolve()
ROOT = next(
    (
        p for p in (HERE, *HERE.parents)
        if (p / "replay_manager").is_dir()
        and (p / "parser").is_dir()
    ),
    HERE,
)

# Broad discovery pattern: don't restrict source/destination yet.
# .+ allows card names containing quotes, e.g. Spirit Message "I".
PATTERN = re.compile(
    r'^Placed\s+"(.+)"\s+from\s+(.+?)\s+to\s+(.+?)\s*$'
)

formats = defaultdict(list)
raw_logs = defaultdict(list)
folders = Counter()
issues = []
records = []  # Complete events remain available for further inspection.
json_files = replay_files = 0

for path in sorted(ROOT.rglob("*.json")):
    if any(
        part in {
            ".git", ".venv", "venv",
            "node_modules", ".ipynb_checkpoints",
        }
        for part in path.relative_to(ROOT).parts
    ):
        continue

    json_files += 1
    relative = str(path.relative_to(ROOT))

    try:
        data = json.loads(path.read_text(encoding="utf-8-sig"))
    except (OSError, ValueError) as exc:
        issues.append(f"{relative}: {exc}")
        continue

    if not isinstance(data, dict) or "plays" not in data:
        continue

    if not isinstance(data["plays"], list):
        issues.append(f"{relative}: plays is not a list")
        continue

    replay_files += 1
    folder = str(path.parent.relative_to(ROOT))
    folders.setdefault(folder, 0)

    for index, play in enumerate(data["plays"]):
        location = f"{relative}, plays[{index}]"  # Zero-based index.

        if not isinstance(play, dict):
            issues.append(f"{location}: play is not an object")
            continue

        if play.get("play") != "To ST":
            continue

        folders[folder] += 1
        records.append((location, play))

        log = play.get("log")
        if not isinstance(log, dict):
            issues.append(f"{location}: missing or invalid log")
            continue

        payload = play.get("card")
        if not isinstance(payload, dict):
            issues.append(f"{location}: missing card object")
            payload = {}

        if payload.get("object_id") is None:
            issues.append(f"{location}: missing card.object_id")

        found_log = False

        # Inspect BOTH versions, rather than hiding one behind a fallback.
        for visibility in ("private_log", "public_log"):
            text = log.get(visibility)
            if not text:
                continue

            found_log = True

            if not isinstance(text, str):
                issues.append(f"{location}: {visibility} is not text")
                continue

            raw_logs[(visibility, text)].append(location)
            match = PATTERN.fullmatch(text.strip())

            if not match:
                issues.append(
                    f"{location}: unmatched {visibility}: {text!r}"
                )
                continue

            name, source, destination = match.groups()

            # Group hand positions while retaining every original log.
            normalized_source = re.sub(
                r"\(\d+/\d+\)", "(#/#)", source
            )
            formats[
                (visibility, normalized_source, destination)
            ].append(location)

            if (
                source not in {"GY", "Deck"}
                and not re.fullmatch(
                    r"hand(?:\s+\(\d+/\d+\))?", source
                )
            ):
                issues.append(
                    f"{location}: handler does not support "
                    f"source {source!r}"
                )

            if not re.fullmatch(r"S-[1-5]", destination):
                issues.append(
                    f"{location}: handler does not support "
                    f"destination {destination!r}"
                )

            if payload.get("name") and payload["name"] != name:
                issues.append(
                    f"{location}: log/card name mismatch"
                )

        if not found_log:
            issues.append(
                f"{location}: no nonempty private/public log"
            )

print(f"Root: {ROOT}")
print(
    f"{json_files} JSON files; {replay_files} replay files; "
    f"{len(records)} To ST events"
)

print("\nEvents by folder:")
for folder, count in sorted(folders.items()):
    print(f"  {folder}: {count}")

print("\nFormats (private/public logs counted separately):")
for (visibility, source, destination), locations in sorted(
    formats.items()
):
    print(
        f'  {len(locations):3} | {visibility} | '
        f'Placed "<name>" from {source} to {destination}'
    )

print("\nEvery distinct raw log, with all file/index locations:")
for (visibility, text), locations in sorted(raw_logs.items()):
    print(f"\n  [{visibility}] {text}")
    for location in locations:
        print(f"    {location}")

print(f"\nIssues requiring inspection: {len(issues)}")
for issue in issues:
    print("  " + issue)
```
